# ESA 子集异常与实验缺失分布分析

## tl;dr

- 原始观测掩码中空值为 **0**；ESA 标注异常不是原始缺测。
- ESA 异常/稀有事件共 **95,657** 个通道-时间单元，占 **1.81%**。
- 至少一个通道异常的时间点为 **16,424**，占 **1.87%**。
- 当前 CSDI 加载器只使用全通道正常窗口；评估缺失由 10%/50%/90% 随机掩蔽另外生成。

## Context & Methods

### Key Assumptions

- ESA 标签码 1、2 分别视为异常和稀有事件，均归入“不干净/不可训练”位置。
- 异常率以 878,400 × 6 个通道-时间单元为分母；时间点影响率以 878,400 个时间戳为分母。
- 连续异常段以 30 秒相邻标签连续为定义。人工缺失统计按重叠窗口中的评估单元计数。

## Takeaways

1. ESA 标签异常与 CSDI 人工缺失是两层不同机制：前者用于排除窗口，后者才是模型评价目标。
2. 异常分布并不均匀，验证集没有 ESA 标签异常，测试集异常集中在 10 月。
3. 大多数受影响时间点六个通道同步异常，若按单元数报告必须同时给出去重后的受影响时间点数。
4. 人工掩蔽在通道和窗口位置上近似均匀，但 10% 和 90% 因每窗整数取整分别对应 10.069% 和 89.931%。
